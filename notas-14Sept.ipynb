### Derivada del modelo logístico

Si tomamos la derivada del modelo logístico:

$$C'(t) = rC(t)\left[1 - \frac{C(t)}{K}\right]$$

donde:
- $r$ = tasa de crecimiento
- $K$ = capacidad de carga

**Para casos acumulados:** $C(t)$ crece de forma sigmoide en función de $t$.

**Para casos diarios:** la curva tiene forma de campana (pico y luego decrecimiento).

### Estimación de parámetros

**Antes qué hacíamos:**
Estimar $r$ y $K$ mediante el método de mínimos cuadrados.

**Ahora qué haremos:**
Encontrar $r$ y $K$ minimizando una función diferente:

$$\min_{r,K} \prod_{i=1}^{n} \hat{y}_i^{\,y_i} \exp(-\hat{y}_i)$$

donde:
- $\hat{y}_i$ = solución a través del modelo
- $y_i$ = datos


### ¿De dónde sale esta función?

Suponiendo que el número de nuevos casos al tiempo $i$, $y_i$, sigue una distribución **Poisson**, con número esperado de casos dado por $\hat{y}_i$.

En el tiempo $t_i$: $\hat{y}_i(r,K) \to$ punto promedio de la distribución de Poisson.

**Variable aleatoria:** Una variable aleatoria $X$ es una cantidad que podemos medir o contar y que tiene probabilidades asociadas a sus valores. 

**Variable aleatoria Poisson:** mide el número de ocurrencias de un evento en un intervalo de tiempo. Estas variables toman valores en $\{0, 1, 2, 3, \dots\}$.

Las probabilidades asociadas al modelo Poisson están dadas por:

$$P(X = a) = \frac{\lambda^a e^{-\lambda}}{a!}, \qquad a = 0, 1, 2, \dots$$

donde $\lambda$ se conoce como **parámetro de intensidad**, e indica el número promedio de casos por unidad de tiempo.


### Función de verosimilitud

Si tenemos una secuencia de casos diarios, digamos $d_1, d_2, \dots, d_n$, podemos preguntarnos: ¿qué tan probable es observar esa secuencia?

$$P(X_1 = d_1, X_2 = d_2, \dots, X_n = d_n) \to \text{Función de Verosimilitud}$$

("qué tan probable es observar mis datos")

Esto se puede calcular:

$$= \prod_{i=1}^{n} P(X_i = d_i) = \prod_{i=1}^{n} \frac{\lambda^{d_i} e^{-\lambda}}{d_i!} = \frac{\lambda^{\sum d_i} e^{-n\lambda}}{\prod_{i=1}^{n} d_i!} \quad (\star)$$

(queda como función de $\lambda$)

Si maximizamos la ecuación $(\star)$ en términos de $\lambda$, podemos encontrar el número promedio de casos $\lambda$ que hace más probable observar la secuencia $d_1, \dots, d_n$.

Según la ecuación:

$$\hat{y}_i \approx \lambda, \qquad y_i \approx d_i$$

En el caso anterior estamos suponiendo que $\lambda$ es único, pero en realidad se tiene un $\lambda$ distinto para cada día. Más aún, $\lambda(t)$ depende de los parámetros del modelo logístico:

$$\lambda(t; r, K) = C(t; r, K) - C(t-1; r, K)$$
